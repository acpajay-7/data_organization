```python

import os
import re

OUTPUT_FOLDER = "/content/drive/MyDrive/cleaned_data_reincrew/recruitment agencies/recruitment_agencies_49904"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)


if "df10" not in globals():
    raise NameError("df10 is not defined. Load your DataFrame first.")

if "city" not in df10.columns:
    raise ValueError(
        f"City column not found. Available columns: {list(df10.columns)}"
    )

try:
    import openpyxl
except ImportError:
    %pip install -q openpyxl
    import openpyxl

data = df10.copy()


data["city"] = (
    data["city"]
    .astype("string")
    .str.lower()
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)


data.loc[data["city"] == "", "city"] = pd.NA

city_mapping = {
    "bengaluru/ bangalore": "bangalore",
    "karnataka(bangalore)": "bangalore",
    "mysoru/ mysore": "mysore",
    "noida/ greater noida": "noida",
    "panjim/ panaji": "panaji",
    "cochin/ kochi/ ernakulam": "kochi",
    "calicut/ kozhikode": "kozhikode",
    "hyderabad/ secunderabad": "hyderabad",
    "tamil nadu(chennai)": "chennai",
    "tamil nadu(madurai)": "madurai",
    "tamil nadu(cuddalore)": "cuddalore",
    "goa(margoa)": "margao",
    "goa(madgao)": "margao",
    "goa(madgaon)": "margao",
    "valsad -vapi": "vapi",
    "maharashtra(new panvel)": "navi mumbai",
    "andhra pradesh(elluru)": "eluru",
    "andhra pradesh(eluru)": "eluru",
    "andhra pradesh(visakhapatnam)": "visakhapatnam",
    "gujarat(ahemdabad)": "ahmedabad",
    "rajasthan(bhilwara)": "bhilwara",
    "rajasthan(jaipur)": "jaipur",
    "rajasthan(sikar)": "sikar",
    "rajasthan(bhiwadi)": "bhiwadi",
    "kerala(cochin)": "kochi",
    "kerala(calicut)": "kozhikode",
    "andhra pradesh(a.p.)": "andhra pradesh",
    "andhra pradesh(andhra pradesh)": "andhra pradesh",
    "jammu &amp; kashmir-": "jammu & kashmir",
    "orissa": "odisha",
    "orissa(orissa)": "odisha",
    "orissa(bhubaneswar)": "bhubaneswar",
    "bhubaneshwar": "bhubaneswar",
}

data["city"] = data["city"].replace(city_mapping)

missing_city_rows = int(data["city"].isna().sum())
data = data.dropna(subset=["city"]).copy()

def safe_filename(city):
    name = str(city).strip()
    name = re.sub(r'[<>:"/\\|?*\x00-\x1f]', "_", name)
    name = re.sub(r"\s+", "_", name)
    name = name.strip("._ ")
    return name or "unknown_city"


summary = []
used_filenames = set()

for city, city_df in data.groupby("city", sort=True):

    base_name = safe_filename(city)
    filename = base_name
    counter = 2


    while filename.lower() in used_filenames:
        filename = f"{base_name}_{counter}"
        counter += 1

    used_filenames.add(filename.lower())

    file_path = os.path.join(
        OUTPUT_FOLDER,
        f"{filename}.xlsx"
    )


    city_df.to_excel(
        file_path,
        index=False,
        engine="openpyxl"
    )

    summary.append({
        "city": city,
        "excel_file": f"{filename}.xlsx",
        "total_rows": len(city_df)
    })

    print(f"Saved: {filename}.xlsx | Rows: {len(city_df)}")

summary_df = pd.DataFrame(summary)

summary_path = os.path.join(
    OUTPUT_FOLDER,
    "city_wise_summary.xlsx"
)

summary_df.to_excel(
    summary_path,
    index=False,
    engine="openpyxl"
)

print("\n" + "=" * 50)
print("ALL CITY-WISE EXCEL FILES SAVED!")
print("=" * 50)
print("Drive folder:", OUTPUT_FOLDER)
print("Total city Excel files:", len(summary_df))
print("Total exported rows:", int(summary_df["total_rows"].sum()))
print("Rows skipped due to missing city:", missing_city_rows)
print("Summary Excel:", summary_path)

display(summary_df)


```
